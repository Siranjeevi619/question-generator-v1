```python
from transformers import pipeline
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import PromptTemplate
```

```python
pipe = pipeline(
    "text2text-generation",
    model="google/flan-t5-base",
    max_new_tokens=512,
    do_sample=False
)

llm = HuggingFacePipeline(pipeline=pipe)
```

```python
question_prompt = PromptTemplate(
    input_variables=["topic", "level"],
    template="""
You are an expert educator.

Generate several clear and simple {level}-level questions
about the topic "{topic}".

Rules:
- Each question must be on a new line
- Do NOT include answers
- Do NOT include explanations
"""
)
```

```python
chain = question_prompt | llm
```

```python
def generate_questions(topic, level="beginner", count=5):
    raw = chain.invoke({
        "topic": topic,
        "level": level
    })

    # split & clean
    questions = [
        q.strip("-•1234567890. ").strip()
        for q in raw.split("\n")
        if q.strip()
    ]

    # enforce exact count
    return questions[:count]
```

```python
questions = generate_questions("Java Threads", "beginner", 5)

for i, q in enumerate(questions, 1):
    print(f"{i}. {q}")
```
