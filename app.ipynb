```python
from dotenv import load_dotenv
import os
```

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()
```

```python
llm = ChatGroq(
    api_key=os.getenv("GROQ_API_KEY"),
    model="llama-3.1-8b-instant",
    temperature=0.7
)
```

```python
from prompts import question_prompt
```

```python
chain = question_prompt | llm
```

```python
def generate_questions(topic, level, count):
    response = chain.invoke({
        "topic": topic,
        "level": level,
        "count": count
    })
    return response.content
```

```python
print(generate_questions("react", "hard", 5))
```

```python
import streamlit as st
```

```python
st.set_page_config(page_title="AI Question Generation",layout="centered")
```

```python
st.title("Question Generator")
```

```python
st.write("Generate interview questions using LangChain + Groq")
```

```python
topic = st.text_input("Topic", value="JVM")
```

```python
level = st.selectbox(
    "Difficulty Level",
    ["Easy", "Medium", "Hard"]
)
```

```python
count = st.number_input(
    "Number of Questions",
    min_value=1,
    max_value=10,
    value=5
)
```

```python
if st.button("Generate Questions"):
    with st.spinner("Generating questions..."):
        result = generate_questions(topic, level, count)

    st.success("Questions generated successfully!")
    st.text_area(
        "Generated Questions",
        result,
        height=300
    )
```
